```python
import pandas as pd
import sys
import os
```

```python
MODULE_A_PATH = rf"C:\Users\haika\Desktop\May_Research\MAC Model for may dataset\ModuleA\Output_Module_A_w_merge_Cora_integrate.csv"
MODULE_C_PATH = rf"C:\Users\haika\Desktop\May_Research\MAC Model for may dataset\ModuleC\Output_Module_C_Babs_Bscat_SSA_HANYANG+CORA.csv"
MODULE_D_PATH = rf"C:\Users\haika\Desktop\May_Research\MAC Model for may dataset\ModuleD\Output_Module_D.csv"
ORGANIZATION_CAMPAIGN_PATH = rf"C:\Users\haika\Desktop\May_Research\MAC Model for may dataset\ModuleA\master_restricted_datetimefixed.csv"

df_module_a = pd.read_csv(MODULE_A_PATH)
df_module_c = pd.read_csv(MODULE_C_PATH)
df_module_d = pd.read_csv(MODULE_D_PATH)
df_org_campaign = pd.read_csv(ORGANIZATION_CAMPAIGN_PATH)
df_org_campaign = df_org_campaign[["Organization", "Campaign", "Date", "UTC", "Latitude", "Longitude", "Altitude", "Temperature", "Rel_humidity", "Pressure", 'bin1', 'bin2', 'bin3', 'bin4', 'bin5', 'bin6', 'bin7', 'bin8', 'bin9', 'bin10', 'bin11', 'bin12', 'bin13', 'bin14', 'bin15']]

print(rf"Module A: {len(df_module_a)} rows, Module C: {len(df_module_c)} rows, Module D: {len(df_module_d)} rows and Organization Campaign: {len(df_org_campaign)} rows")
```

```python
# Combine all three DataFrames horizontally
combined_df = pd.concat([df_org_campaign, df_module_a, df_module_c, df_module_d], axis=1)

print(f"Before removing duplicates: {len(combined_df)} rows, {len(combined_df.columns)} columns")
print(f"All column names: {list(combined_df.columns)}")

# Find duplicate "datetime" columns
datetime_columns = [col for col in combined_df.columns if col == 'datetime']
print(f"\nFound {len(datetime_columns)} 'datetime' columns")

if len(datetime_columns) > 1:
    print(f"Duplicate 'datetime' columns found: {datetime_columns}")
    
    # Use pandas built-in method to remove duplicate columns (keeps first occurrence)
    print("Removing duplicate columns...")
    combined_df = combined_df.loc[:, ~combined_df.columns.duplicated()]
    
    # Verify removal
    datetime_columns_after = [col for col in combined_df.columns if col == 'datetime']
    print(f"Removed {len(datetime_columns) - len(datetime_columns_after)} duplicate 'datetime' columns")
    
else:
    print("No duplicate 'datetime' columns found")

print(f"\nAfter removing duplicates: {len(combined_df)} rows, {len(combined_df.columns)} columns")
print(f"Final column names: {list(combined_df.columns)}")

# Verify there's only one datetime column now
final_datetime_count = sum(1 for col in combined_df.columns if col == 'datetime')
print(f"Final count of 'datetime' columns: {final_datetime_count}")

# Double-check: if there are still duplicates, force remove them manually
if final_datetime_count > 1:
    print(f"WARNING: Still {final_datetime_count} datetime columns found. Manually removing duplicates...")
    
    # Get column positions of datetime columns
    datetime_positions = [i for i, col in enumerate(combined_df.columns) if col == 'datetime']
    print(f"Datetime column positions: {datetime_positions}")
    
    # Keep only columns that are NOT duplicate datetime columns (except the first one)
    columns_to_keep = []
    datetime_kept = False
    
    for i, col in enumerate(combined_df.columns):
        if col == 'datetime':
            if not datetime_kept:
                columns_to_keep.append(i)
                datetime_kept = True
                print(f"Keeping first 'datetime' column at position {i}")
            else:
                print(f"Dropping duplicate 'datetime' column at position {i}")
        else:
            columns_to_keep.append(i)
    
    # Select columns by position
    combined_df = combined_df.iloc[:, columns_to_keep]
    
    # Final verification
    final_datetime_count = sum(1 for col in combined_df.columns if col == 'datetime')
    print(f"Final count of 'datetime' columns after manual removal: {final_datetime_count}")

combined_df.to_csv(rf"C:\Users\haika\Desktop\May_Research\MAC Model for may dataset\Module_combined\MAC_NASA_DOE_3Bins_weather_originalbins.csv", index=False)

print(f"\nCombined DataFrame saved: {len(combined_df)} rows, {len(combined_df.columns)} columns")
print(f"Final column count verification: {len(combined_df.columns)} columns")
print(f"Datetime columns in final output: {[col for col in combined_df.columns if col == 'datetime']}")
```
