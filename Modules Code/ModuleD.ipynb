```python
#Assign necessary values

ABS_NEW_WL = 870
```

```python
#Read in necessary files, BC_Mass.csv, Output_Module_C.

df_bcmass = pd.read_csv(rf'C:\Users\haika\Desktop\May_Research\MAC Model for may dataset\ModuleD\BC_mass.csv')
df_moduleC = pd.read_csv(rf"C:\Users\haika\Desktop\May_Research\MAC Model for may dataset\ModuleC\Output_Module_3_Combined.csv")
```

```python
#
# Calculate babs_870 
#

bcmass = df_bcmass['BC_Mass'].tolist()
SAE_three_lambda = df_moduleC['SAE'].tolist()
SAE_constant = df_moduleC['SAE_constant'].tolist()
AAE_three_lambda = df_moduleC['AAE'].tolist()
AAE_constant = df_moduleC['AAE_constant'].tolist()

#List to store babs_870 values
babs_new_wl = [0.0] * len(bcmass)

#calculate babs_870 using the formula
babs_new_wl = [
    AAE_constant[i] * np.power(ABS_NEW_WL, -AAE_three_lambda[i])
    for i in range(len(bcmass))
]

#
# Calculate MAC_bc
#

# MAC_bc = [
#     babs_new_wl[i] / bcmass[i] if bcmass[i] != 0 else np.nan
#     for i in range(len(bcmass))
# ]

# Unit conversion factors
Mm_to_m = 1e-6  # Convert Mm^-1 to m^-1
ng_to_g = 1e-9  # Convert ng/m^3 to g/m^3
MAC_bc = [
    (babs_new_wl[i] * Mm_to_m) / (bcmass[i] * ng_to_g) if bcmass[i] != 0 else np.nan
    for i in range(len(bcmass))
]



#result df
df_moduleD = pd.DataFrame()

#date time column added
dateparse = lambda x: datetime.strptime(x, '%m/%d/%Y %H:%M:%S')
datetimedf = pd.read_csv(rf"C:\Users\haika\Desktop\May_Research\MAC Model for may dataset\ModuleA\datetime.csv", parse_dates=['datetime_all'], date_parser=dateparse)['datetime_all']
df_moduleD["datetime"] = datetimedf

#Add babs_870 and MAC_bc to the df
df_moduleD["babs_870"] = babs_new_wl
df_moduleD["MAC_bc"] = MAC_bc
```

```python
#output to csv
df_moduleD.to_csv(rf"C:\Users\haika\Desktop\May_Research\MAC Model for may dataset\ModuleD\Output_Module_D_ZENODO.csv", index=False)
```
