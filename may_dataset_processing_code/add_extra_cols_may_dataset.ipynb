```python
import pandas as pd
import icartt
import os
import warnings
import re
from datetime import datetime
import csv
from datetime import datetime, timedelta
from netCDF4 import Dataset
import numpy as np
from scipy import stats
import glob
from math import pi
import ast
```

```python
MAC_DATASET_PATH = rf"C:\Users\haika\Desktop\May_Research\MAC Model for may dataset\Module_combined\MAC_NASA_DOE_3Bins_weather_originalbins.csv"
RESTRICTED_COMBINED_PATH = rf"C:\Users\haika\Desktop\May_Research\may_datasets\restricted_combined\master_restricted.csv"
COMPREHENSIVE_COMBINED_PATH = rf"C:\Users\haika\Desktop\May_Research\may_datasets\comprehensive_combined\master_comprehensive.csv"

OUTPUT_COMPREHENSIVE_PATH = rf"C:\Users\haika\Desktop\May_Research\may_datasets\comprehensive_combined\master_comprehensive.csv"
OUTPUT_RESTRICTED_PATH = rf"C:\Users\haika\Desktop\May_Research\may_datasets\restricted_combined\master_restricted.csv"


df_mac = pd.read_csv(MAC_DATASET_PATH)
df_restricted = pd.read_csv(RESTRICTED_COMBINED_PATH)
df_comprehensive = pd.read_csv(COMPREHENSIVE_COMBINED_PATH)
```

```python
# Define the columns to merge on and add
COLS_TO_COMBINE_WITH = ['Organization', 'Campaign', 'Date', 'UTC']
COLS_TO_ADD = ["SAE", "SAE_constant", "AAE", "AAE_constant", "SSA_red", "SSA_blue", "SSA_green"]

# Create a subset of df_mac with only the columns needed for merging
df_mac_subset = df_mac[COLS_TO_COMBINE_WITH + COLS_TO_ADD]

# Merge df_mac columns into df_restricted
df_restricted_merged = df_restricted.merge(
    df_mac_subset, 
    on=COLS_TO_COMBINE_WITH, 
    how='left'
)

# Merge df_mac columns into df_comprehensive
df_comprehensive_merged = df_comprehensive.merge(
    df_mac_subset, 
    on=COLS_TO_COMBINE_WITH, 
    how='left'
)

# Save the merged dataframes
df_restricted_merged.to_csv(OUTPUT_RESTRICTED_PATH, index=False)
df_comprehensive_merged.to_csv(OUTPUT_COMPREHENSIVE_PATH, index=False)

print("Merge completed successfully!")
print(f"Restricted dataset saved to: {OUTPUT_RESTRICTED_PATH}")
print(f"Comprehensive dataset saved to: {OUTPUT_COMPREHENSIVE_PATH}")

# Optional: Print merge statistics
print(f"\nMerge statistics:")
print(f"Original restricted rows: {len(df_restricted)}")
print(f"Merged restricted rows: {len(df_restricted_merged)}")
print(f"Original comprehensive rows: {len(df_comprehensive)}")
print(f"Merged comprehensive rows: {len(df_comprehensive_merged)}")

# Check for any rows that didn't get matched
restricted_nulls = df_restricted_merged[COLS_TO_ADD].isnull().sum()
comprehensive_nulls = df_comprehensive_merged[COLS_TO_ADD].isnull().sum()

print(f"\nNull values in merged columns (restricted):")
for col in COLS_TO_ADD:
    print(f"  {col}: {restricted_nulls[col]}")

print(f"\nNull values in merged columns (comprehensive):")
for col in COLS_TO_ADD:
    print(f"  {col}: {comprehensive_nulls[col]}")
```
