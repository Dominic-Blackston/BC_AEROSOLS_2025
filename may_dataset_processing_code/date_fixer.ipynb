```python
import pandas as pd
import re

path_to_fix = rf"C:\Users\haika\Desktop\May_Research\may_datasets\comprehensive_combined\NOAA_comprehensive.csv"

# Read the existing CSV
df = pd.read_csv(path_to_fix)

# Function to check if a date is already in YYYYMMDD format
def is_yyyymmdd_format(date_str):
    """Check if string matches YYYYMMDD format (8 digits)"""
    if pd.isna(date_str):
        return True  # Skip NaN values
    date_str = str(date_str).strip()
    return bool(re.match(r'^\d{8}$', date_str))

# Check which dates need conversion
print("Checking date formats...")
dates_to_convert = ~df['Date'].apply(is_yyyymmdd_format)
num_to_convert = dates_to_convert.sum()

print(f"Total rows: {len(df)}")
print(f"Rows needing conversion: {num_to_convert}")
print(f"Rows already in YYYYMMDD format: {len(df) - num_to_convert}")

if num_to_convert > 0:
    print(f"\nSample dates that need conversion:")
    sample_bad_dates = df.loc[dates_to_convert, 'Date'].head(5).tolist()
    print(sample_bad_dates)
    
    # Only convert the dates that need it
    print("\nConverting non-YYYYMMDD dates...")
    df.loc[dates_to_convert, 'Date'] = pd.to_datetime(
        df.loc[dates_to_convert, 'Date'], 
        format='mixed'  # Handle multiple input formats
    ).dt.strftime('%Y%m%d')
    
    # Save back to CSV
    df.to_csv(path_to_fix, index=False)
    print("Date format fixed and saved!")
else:
    print("All dates are already in YYYYMMDD format - no changes needed!")

# Show final sample
print(f"\nFinal sample dates: {df['Date'].head().tolist()}")

# Verify all dates are now in correct format
all_correct = df['Date'].apply(is_yyyymmdd_format).all()
print(f"All dates now in YYYYMMDD format: {all_correct}")
```
