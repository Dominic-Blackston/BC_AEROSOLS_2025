```python
import pandas as pd
PATH = rf"C:\Users\haika\Desktop\May_Research\MAC Machine Learning Model Code\MAC_DATASET_LLOD_FILTERED_V3.csv"
OUTPUT_PATH = rf"C:\Users\haika\Desktop\May_Research\MAC Machine Learning Model Code\MAC_DATASET_LLOD_FILTERED_V3_10campaigns.csv"
# Read the data and get initial campaign info
df = pd.read_csv(PATH)

# Get initial statistics
initial_campaigns = df['Campaign'].unique()
initial_count = len(initial_campaigns)
initial_rows = len(df)

print("BEFORE FILTERING:")
print(f"Unique campaigns: {initial_count}")
print(f"Total rows: {initial_rows}")
print("Campaign list:")
for campaign in sorted(initial_campaigns):
    print(f"  - {campaign}")

# Apply filtering
campaigns_to_drop = ["NAAMES(2015)", "NAAMES(2016)", "NAAMES(2017)"]
df = df[~df['Campaign'].isin(campaigns_to_drop)]

# Get final statistics
final_campaigns = df['Campaign'].unique()
final_count = len(final_campaigns)
final_rows = len(df)

print("\nAFTER FILTERING:")
print(f"Unique campaigns: {final_count}")
print(f"Total rows: {final_rows}")
print("Campaign list:")
for campaign in sorted(final_campaigns):
    print(f"  - {campaign}")

print(f"\nSUMMARY:")
print(f"Campaigns removed: {initial_count - final_count}")
print(f"Rows removed: {initial_rows - final_rows}")
print(f"Percentage of data retained: {(final_rows/initial_rows)*100:.1f}%")

df.to_csv(OUTPUT_PATH, index=False)
```
