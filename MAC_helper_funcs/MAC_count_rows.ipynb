```python
import pandas as pd
import numpy as np

# Read in the LLOD filtered MAC dataset
df = pd.read_csv(rf"C:\Users\haika\Desktop\May_Research\MAC Machine Learning Model Code\MAC_DATASET_LLOD_FILTERED_V3_10campaigns.csv")

print("CAMPAIGN AND ORGANIZATION ROW ANALYSIS - LLOD FILTERED DATASET")
print("=" * 70)

# Calculate total rows in the dataset
total_rows = len(df)
print(f"Total rows in filtered dataset: {total_rows:,}")

# CAMPAIGN ANALYSIS
print(f"\n1. ROWS PER CAMPAIGN:")
print("=" * 50)

# Get campaign counts and percentages
campaign_counts = df['Campaign'].value_counts().sort_index()

print(f"{'Campaign':<25} {'Rows':<10} {'Percentage':<10}")
print("-" * 45)

# Calculate and display each campaign's statistics
for campaign, count in campaign_counts.items():
    percentage = (count / total_rows) * 100
    print(f"{campaign:<25} {count:<10,} {percentage:<10.2f}%")

print("-" * 45)
print(f"{'TOTAL':<25} {total_rows:<10,} {'100.00%':<10}")

# ORGANIZATION ANALYSIS
print(f"\n2. ROWS PER ORGANIZATION:")
print("=" * 50)

# Check if Organization column exists
if 'Organization' in df.columns:
    # Get organization counts and percentages
    org_counts = df['Organization'].value_counts().sort_index()
    
    print(f"{'Organization':<25} {'Rows':<10} {'Percentage':<10}")
    print("-" * 45)
    
    # Calculate and display each organization's statistics
    for organization, count in org_counts.items():
        percentage = (count / total_rows) * 100
        print(f"{organization:<25} {count:<10,} {percentage:<10.2f}%")
    
    print("-" * 45)
    print(f"{'TOTAL':<25} {total_rows:<10,} {'100.00%':<10}")
    
    # Organization summary statistics
    print(f"\nOrganization Summary:")
    print(f"- Number of organizations: {len(org_counts)}")
    print(f"- Largest organization: {org_counts.index[0]} ({org_counts.iloc[0]:,} rows, {100*org_counts.iloc[0]/total_rows:.1f}%)")
    print(f"- Smallest organization: {org_counts.index[-1]} ({org_counts.iloc[-1]:,} rows, {100*org_counts.iloc[-1]/total_rows:.1f}%)")
    print(f"- Average rows per organization: {org_counts.mean():.0f}")
    print(f"- Median rows per organization: {org_counts.median():.0f}")
    
    # Show top organizations by size
    print(f"\nTop organizations by number of rows:")
    print("-" * 40)
    for i, (organization, count) in enumerate(org_counts.head().items(), 1):
        percentage = (count / total_rows) * 100
        print(f"{i}. {organization}: {count:,} rows ({percentage:.1f}%)")
else:
    print("Warning: 'Organization' column not found in dataset")

# CAMPAIGN SUMMARY STATISTICS
print(f"\n3. CAMPAIGN SUMMARY STATISTICS:")
print("=" * 50)
print(f"- Number of campaigns: {len(campaign_counts)}")
print(f"- Largest campaign: {campaign_counts.index[0]} ({campaign_counts.iloc[0]:,} rows, {100*campaign_counts.iloc[0]/total_rows:.1f}%)")
print(f"- Smallest campaign: {campaign_counts.index[-1]} ({campaign_counts.iloc[-1]:,} rows, {100*campaign_counts.iloc[-1]/total_rows:.1f}%)")
print(f"- Average rows per campaign: {campaign_counts.mean():.0f}")
print(f"- Median rows per campaign: {campaign_counts.median():.0f}")

# Show top 5 campaigns by size
print(f"\nTop 5 campaigns by number of rows:")
print("-" * 40)
for i, (campaign, count) in enumerate(campaign_counts.head(5).items(), 1):
    percentage = (count / total_rows) * 100
    print(f"{i}. {campaign}: {count:,} rows ({percentage:.1f}%)")
```
